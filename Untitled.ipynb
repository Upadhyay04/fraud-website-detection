```python
import subprocess

# Start Flask app
flask_process = subprocess.Popen(["python", "app.py"])

print("🚀 Flask app launched. It's running at http://127.0.0.1:5000/")
```

```python
import requests

test_url = "http://example.com"  # You can try other URLs too

res = requests.post("http://127.0.0.1:5000/check", json={"url": test_url})

print("Status Code:", res.status_code)
print("Response:", res.json())
```

```python
status = flask_process.poll()
if status is None:
    print("🚀 Flask server is still running.")
else:
    print(f"❌ Flask server exited with code {status}.")
```

```python
# Stop the Flask app
flask_process.terminate()
print("🛑 Flask app stopped.")
```
